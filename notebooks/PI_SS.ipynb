```python
def preprocess(df, tahun):

    # Ambil header penting
    kategori = pd.Series(df.iloc[1].values).ffill().values
    bulan = df.iloc[3].values

    # Buang header
    df = df.iloc[4:].reset_index(drop=True)

    # Rename kolom pertama
    df = df.rename(columns={df.columns[0]: "Bandara"})

    # Samakan panjang
    kategori = kategori[:len(df.columns)]
    bulan = bulan[:len(df.columns)]

    # Buang kolom Tahunan
    mask = [b != "Tahunan" for b in bulan]

    df = df.loc[:, mask]
    kategori = kategori[mask]
    bulan = bulan[mask]

    # Buat nama kolom
    kolom_baru = ["Bandara"]

    for k, b in zip(kategori[1:], bulan[1:]):
        kolom_baru.append(f"{k}_{b}")

    df.columns = kolom_baru

    # Ambil hanya 5 bandara utama
    bandara_utama = [
        "Kualanamu-Medan",
        "Soekarno Hatta-Jakarta",
        "Juanda-Surabaya",
        "Hasanudin-Makassar",
        "Ngurah Rai-Bali"
    ]

    df = df[df["Bandara"].isin(bandara_utama)]

    # Tambah tahun
    df["Tahun"] = tahun

    return df
```

```python
# GABUNG SEMUA TAHUN
all_data = []

for tahun in range(2019, 2026):

    file = f"penumpang_bandara_{tahun}.csv"

    df = pd.read_csv(file)

    df = preprocess(df, tahun)

    all_data.append(df)

# SATUKAN
final_df = pd.concat(all_data, ignore_index=True)

final_df.head()
```

```python
final_df["Tahun"].unique()

final_df.shape
```

```python
pd.set_option('display.max_columns', None)

print(final_df.head())
```

```python
df_long = final_df.melt(
    id_vars=["Bandara", "Tahun"],
    var_name="Kategori_Bulan",
    value_name="Jumlah"
)

df_long[["Kategori", "Bulan"]] = df_long["Kategori_Bulan"].str.split("_", expand=True)

df_long = df_long.drop(columns=["Kategori_Bulan"])

df_long.head()
```

```python
df_pivot = df_long.pivot_table(
    index=["Bandara", "Tahun", "Bulan"],
    columns="Kategori",
    values="Jumlah",
    aggfunc="sum"
).reset_index()

df_pivot.head()
```

```python
df_pivot.to_csv("data_pivot_bandara.csv", index=False)
```
